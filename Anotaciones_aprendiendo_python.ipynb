## 📘 Clase 0: Los símbolos de Python y para qué sirve cada uno

Antes de empezar, conoce los símbolos que vas a ver todo el tiempo. No tienes que memorizarlos hoy: vuelve a esta tabla cada vez que tengas duda.

### 1. Operadores matemáticos (hacen cuentas)

| Símbolo | Nombre | Para qué sirve | Ejemplo | Resultado |
|---|---|---|---|---|
| `+` | Suma | Suma números (y **une** textos) | `7 + 2` | `9` |
| `-` | Resta | Resta números (y hace negativos: `-5`) | `7 - 2` | `5` |
| `*` | Multiplicación | Multiplica. **No se usa `x`** para multiplicar | `7 * 2` | `14` |
| `/` | División | Divide. **Siempre da decimal** (`float`) | `7 / 2` | `3.5` |
| `//` | División entera | Divide y **quita los decimales** (redondea hacia abajo) | `7 // 2` | `3` |
| `%` | Módulo (residuo) | Da **lo que sobra** de una división | `7 % 2` | `1` |
| `**` | Potencia | Eleva a una potencia. **No se usa `^`** | `7 ** 2` | `49` |

```python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
```

### 🔍 Los que más confunden: `/`, `//` y `%`

Piensa en repartir **17 dulces entre 5 niños**:
- `17 / 5` = **3.4** → la división "exacta", con decimales.
- `17 // 5` = **3** → a cada niño le tocan **3 dulces completos**.
- `17 % 5` = **2** → **sobran 2 dulces** (el residuo).

Comprobación: 5 niños × 3 dulces + 2 que sobran = 17. ✅

```python
dulces = 17
ninos = 5
print("División normal:", dulces / ninos)
print("A cada niño le tocan:", dulces // ninos)
print("Sobran:", dulces % ninos)
```

**¿Para qué sirve `%` en la vida real?** Muchísimo:
- **Saber si un número es par:** si `n % 2` da `0`, es par; si da `1`, es impar.
- **Saber si un número es divisible entre otro:** si `n % 3 == 0`, es múltiplo de 3.
- **Sacar el último dígito de un número:** `4827 % 10` da `7`.
- **Convertir minutos a horas y minutos:** `135 // 60` = 2 horas y `135 % 60` = 15 minutos.

```python
print(10 % 2)     # 0 → 10 es par
print(7 % 2)      # 1 → 7 es impar
print(4827 % 10)  # 7 → último dígito

minutos = 135
print(minutos // 60, "horas y", minutos % 60, "minutos")
```

### 🔍 La potencia `**` (y raíces)
- `2 ** 3` = 2 × 2 × 2 = **8**
- `5 ** 2` = **25** (5 al cuadrado)
- Truco: elevar a `0.5` es sacar **raíz cuadrada**: `16 ** 0.5` = **4.0**

```python
print(2 ** 3)
print(5 ** 2)
print(16 ** 0.5)
print(10 ** 6)    # un millón
```

### 🔍 El orden de las operaciones (jerarquía)
Python respeta el mismo orden que en matemáticas:
1. `( )` paréntesis
2. `**` potencias
3. `*`, `/`, `//`, `%` (de izquierda a derecha)
4. `+`, `-` (de izquierda a derecha)

Si tienes duda, **usa paréntesis**: nunca estorban.

```python
print(2 + 3 * 4)       # 14: primero 3 * 4
print((2 + 3) * 4)     # 20: primero el paréntesis
print(2 * 3 ** 2)      # 18: primero la potencia (3 ** 2 = 9)
print((2 * 3) ** 2)    # 36
```

### 2. Operadores de asignación (guardan valores)

| Símbolo | Para qué sirve | Ejemplo | Equivale a |
|---|---|---|---|
| `=` | **Guarda** un valor en una variable (¡no es "igual"!) | `x = 5` | — |
| `+=` | Suma y guarda | `x += 3` | `x = x + 3` |
| `-=` | Resta y guarda | `x -= 3` | `x = x - 3` |
| `*=` | Multiplica y guarda | `x *= 3` | `x = x * 3` |
| `/=` | Divide y guarda | `x /= 3` | `x = x / 3` |

```python
x = 10
x += 5     # ahora x vale 15
print(x)
x -= 3     # ahora 12
print(x)
x *= 2     # ahora 24
print(x)
x /= 4     # ahora 6.0 (la división da decimal)
print(x)
```

### 3. Operadores de comparación (hacen preguntas → `True` o `False`)
Los vemos a fondo en la clase 5.

| Símbolo | Pregunta | Ejemplo | Resultado |
|---|---|---|---|
| `==` | ¿Es igual? (**dos** signos de igual) | `5 == 5` | `True` |
| `!=` | ¿Es distinto? | `5 != 3` | `True` |
| `>` | ¿Es mayor? | `5 > 3` | `True` |
| `<` | ¿Es menor? | `5 < 3` | `False` |
| `>=` | ¿Es mayor o igual? | `5 >= 5` | `True` |
| `<=` | ¿Es menor o igual? | `4 <= 3` | `False` |

### 4. Operadores lógicos (combinan preguntas)

| Palabra | Significa | Ejemplo | Resultado |
|---|---|---|---|
| `and` | "y": las dos deben ser verdad | `5 > 3 and 2 > 1` | `True` |
| `or` | "o": basta con que una sea verdad | `5 > 9 or 2 > 1` | `True` |
| `not` | "no": invierte el resultado | `not True` | `False` |

```python
print(5 == 5)
print(5 != 5)
print(5 > 3 and 2 > 1)
print(5 > 9 or 2 > 1)
print(not True)
```

## 📘 Clase 1: Variables

Una variable es una **caja con nombre** donde guardas un dato.

El signo `=` **no** significa "es igual": significa **"guarda en"**.
`edad = 20` se lee: *"guarda 20 en la caja llamada `edad`"*.

```python
edad = 20
edad = edad + 1
print(edad)
```

Así lo lee la computadora, **línea por línea, de arriba hacia abajo**:
1. `edad = 20` → guarda 20 en la caja `edad`.
2. `edad = edad + 1` → **primero** calcula lo de la derecha (20 + 1 = 21), **después** lo guarda en `edad`. El 20 se reemplaza.
3. `print(edad)` → muestra **21**.

> 💡 `print()` muestra en pantalla lo que le pongas entre paréntesis.

**Reglas para los nombres:** sin espacios (usa `_`, como `precio_total`), no pueden empezar con número y distinguen mayúsculas (`Edad` ≠ `edad`). Usa nombres que digan **qué guardan**.

**Buena costumbre:** pon espacios alrededor de `=` y de los operadores (`total = precio * cantidad`). Python no lo exige, pero se lee mucho mejor.

## 📘 Clase 2: Tipos de datos

Cada dato en Python tiene un **tipo**. El tipo le dice a la computadora **qué clase de dato es** y **qué puede hacer con él**.

| Tipo | Qué es | Ejemplos |
|---|---|---|
| `int` | Número **entero** (sin decimales) | `250`, `3`, `-10`, `0` |
| `float` | Número **decimal** | `3.5`, `99.99`, `-0.5`, `2.0` |
| `str` | **Texto** (siempre entre comillas) | `"Hola"`, `'Ana'`, `"250"` |
| `bool` | **Verdadero o falso** | `True`, `False` |

### `int`: enteros
Para contar cosas que no se parten: personas, productos, años.

```python
cantidad = 3
edad = 20
print(cantidad, edad)
```

### `float`: decimales
Para precios, medidas y promedios.

⚠️ En Python los decimales se escriben con **punto**, no con coma: `99.50` ✅, `99,50` ❌.

```python
precio = 99.50
estatura = 1.72
print(precio, estatura)

print(7 / 2)    # 3.5 → la división / SIEMPRE da float
print(6 / 2)    # 3.0 → aunque sea exacta, da float
print(7 // 2)   # 3   → la división // quita los decimales y da int
```

### `str`: texto
Todo lo que va entre comillas es texto, **aunque parezca número**.

```python
nombre = "Ana"
codigo = "250"      # ¡esto es texto, no número!
print(nombre, codigo)
```

### `bool`: verdadero o falso
Es lo que responden las comparaciones (lo vemos en la clase 5).

```python
es_mayor = 20 >= 18
print(es_mayor)
```

### Cómo saber el tipo de algo: `type()`

```python
print(type(250))
print(type(99.5))
print(type("250"))
print(type(True))
```

### ⚠️ Por qué importa el tipo

```python
print(250 + 3)       # 253  → suma de números
print("250" + "3")   # 2503 → ¡une los textos, no suma!
```

| Función | Convierte a | Ejemplo | Resultado |

| `int()` | entero | `int("250")` | `250` |

## 📘 Clase 4: f-strings (mezclar texto y variables)

Un **f-string** es un texto con una **`f` antes de las comillas**. Dentro puedes poner variables o cálculos entre **llaves `{ }`**, y Python los reemplaza por su valor.

```python
nombre = "Ana"
edad = 20
print(f"Me llamo {nombre} y tengo {edad} años")
```

**¿Por qué es tan útil?** Sin f-string tendrías que convertir y unir todo a mano:

```python
print("Tengo " + str(edad) + " años")   # funciona, pero es incómodo
print(f"Tengo {edad} años")             # ✅ mucho más fácil
```

### Puedes hacer cálculos dentro de las llaves

```python
precio = 250
cantidad = 3
print(f"Total: {precio * cantidad}")
print(f"El año que viene tendré {edad + 1} años")
```

### Dar formato a los números
Después del nombre de la variable pones `:` y un código de formato:

| Escribes | Qué hace | Resultado |
|---|---|---|
| `{pi:.2f}` | 2 decimales | `3.14` |
| `{precio:.2f}` | 2 decimales (ideal para dinero) | `250.00` |
| `{grande:,}` | separador de miles | `1,500,000` |
| `{grande:,.2f}` | miles y 2 decimales | `1,500,000.00` |
| `{descuento:.0%}` | como porcentaje | `15%` |

```python
pi = 3.14159265
precio = 250
grande = 1500000
descuento = 0.15

print(f"Pi: {pi:.2f}")
print(f"Precio: ${precio:.2f}")
print(f"Premio: ${grande:,}")
print(f"Premio: ${grande:,.2f}")
print(f"Descuento: {descuento:.0%}")
```

⚠️ **Error típico:** olvidar la `f`. Mira la diferencia:

```python
print("Tengo {edad} años")    # sin f: muestra las llaves tal cual
print(f"Tengo {edad} años")   # ✅ con f
```

## 📘 Clase 5: Comparaciones (`True` / `False`)

Para tomar decisiones, primero la computadora necesita **hacer preguntas** cuya respuesta sea **sí (`True`)** o **no (`False`)**.

| Operador | Pregunta | Ejemplo | Resultado |
|---|---|---|---|
| `==` | ¿es igual? | `5 == 5` | `True` |
| `!=` | ¿es distinto? | `5 != 3` | `True` |
| `>` | ¿es mayor? | `5 > 8` | `False` |
| `<` | ¿es menor? | `5 < 8` | `True` |
| `>=` | ¿mayor o igual? | `5 >= 5` | `True` |
| `<=` | ¿menor o igual? | `9 <= 5` | `False` |

⚠️ **El error más común:** `=` **guarda** un valor y `==` **pregunta** si dos valores son iguales.

```python
total = 750
print(total > 500)     # True: 750 sí es mayor que 500
print(total == 1000)   # False: 750 no es igual a 1000
print(total != 1000)   # True: 750 sí es distinto de 1000
```

### Combinar preguntas: `and`, `or`, `not`
- `and` → `True` solo si **las dos** son verdaderas. *("Tengo boleto **y** tengo INE")*
- `or` → `True` si **al menos una** es verdadera. *("Pago en efectivo **o** con tarjeta")*
- `not` → lo invierte: `True` ↔ `False`.

```python
edad = 20
tiene_ine = True

print(edad >= 18 and tiene_ine)   # True: se cumplen las dos
print(edad >= 21 and tiene_ine)   # False: la primera no se cumple
print(edad >= 21 or tiene_ine)    # True: basta con una
print(not tiene_ine)              # False

# Python permite escribir rangos como en matemáticas:
print(18 <= edad <= 30)           # ¿edad está entre 18 y 30?
```
