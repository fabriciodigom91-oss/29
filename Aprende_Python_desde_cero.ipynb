## 1. La idea clave: cómo "piensa" una computadora

Un programa es una **receta**. La computadora la sigue **al pie de la letra**, **una línea a la vez**, **de arriba hacia abajo**. No adivina lo que quisiste decir: hace exactamente lo que escribiste.

Todo programa, por grande que sea, está hecho de solo **3 ingredientes**:

| Ingrediente | Qué significa | En Python |
|---|---|---|
| **Secuencia** | Hacer una cosa, luego la siguiente, luego la siguiente | Líneas una debajo de otra |
| **Decisión** | "Si pasa esto, haz aquello; si no, haz otra cosa" | `if`, `elif`, `else` |
| **Repetición** | "Haz esto varias veces" | `for`, `while` |

Cuando entiendas cuándo usar cada ingrediente, sabrás programar. Lo demás es vocabulario.

**Ejemplo de la vida real (preparar café):**

```
1. Calentar agua                         ← secuencia
2. Si quieres azúcar: agregar azúcar     ← decisión (if)
3. Por cada invitado: servir una taza    ← repetición con cantidad conocida (for)
4. Mientras el café esté caliente: soplar ← repetición hasta que algo cambie (while)
```

Guarda esa última idea en mente, porque es **la diferencia entre `for` y `while`**:
- **`for`** → "por cada…" → **sabes cuántas veces** (o tienes una lista de cosas que recorrer).
- **`while`** → "mientras…" → **no sabes cuántas veces**; repites hasta que una condición cambie.

## 3. Comparaciones y lógica

Antes de tomar decisiones, necesitamos **preguntas** cuya respuesta sea **sí (`True`)** o **no (`False`)**.

| Operador | Pregunta | Ejemplo | Resultado |
|---|---|---|---|
| `==` | ¿Es igual? | `5 == 5` | `True` |
| `!=` | ¿Es distinto? | `5 != 3` | `True` |
| `>` | ¿Es mayor? | `5 > 3` | `True` |
| `<` | ¿Es menor? | `5 < 3` | `False` |
| `>=` | ¿Mayor o igual? | `5 >= 5` | `True` |
| `<=` | ¿Menor o igual? | `4 <= 3` | `False` |

⚠️ **El error número 1 de los principiantes:** `=` guarda un valor; `==` **pregunta** si son iguales.

````python
x = 10
print(x == 10)
print(x != 10)
print(x > 5)
print(x <= 9)
````

### Combinar preguntas: `and`, `or`, `not`

- `and` → `True` solo si **las dos** son verdaderas. ("Tengo boleto **y** tengo identificación")
- `or` → `True` si **al menos una** es verdadera. ("Pago con efectivo **o** con tarjeta")
- `not` → invierte: `True` ↔ `False`.

````python
edad = 20
tiene_boleto = True

print(edad >= 18 and tiene_boleto)   # True: se cumplen las dos
print(edad >= 21 and tiene_boleto)   # False: la primera no se cumple
print(edad >= 21 or tiene_boleto)    # True: basta con una
print(not tiene_boleto)              # False

# Python permite escribir rangos como en matemáticas:
print(0 <= edad <= 100)              # ¿está edad entre 0 y 100?
````

## 4. Decisiones: `if`, `elif`, `else`

### La estructura

```python
if condición:
    # se ejecuta SOLO si la condición es True
elif otra_condición:
    # se revisa SOLO si lo anterior fue False
else:
    # se ejecuta si NINGUNA de las anteriores fue True
```

**Reglas importantes:**
1. Después de la condición van **dos puntos** `:`.
2. Lo que pertenece al `if` va **indentado** (con 4 espacios o Tab). La indentación **no es decoración**: le dice a Python qué líneas están "dentro" del `if`.
3. `elif` y `else` son **opcionales**. Puedes tener un `if` solo, un `if` + `else`, o un `if` + varios `elif` + `else`.
4. **Solo se ejecuta UNA rama**: la primera cuya condición sea `True`. Las demás se saltan.

### 4.1 `if` solo: "haz esto solo si…"

````python
temperatura = 35

if temperatura > 30:
    print("Hace calor, toma agua.")    # dentro del if

print("Este mensaje sale siempre.")    # fuera del if (sin indentar)
````

Cambia `temperatura` a `20` y ejecuta otra vez. El primer mensaje desaparece; el segundo no.

### 4.2 `if` + `else`: "una cosa u otra" (exactamente 2 caminos)

````python
numero = 7

if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
````

### 4.3 `if` + `elif` + `else`: "varias opciones, elige solo una" (3 o más caminos)

````python
nota = 85

if nota >= 90:
    print("Excelente")
elif nota >= 80:
    print("Muy bien")
elif nota >= 70:
    print("Bien")
elif nota >= 60:
    print("Suficiente")
else:
    print("Reprobado")
````

**¿Por qué con 85 sale solo "Muy bien" y no también "Bien" y "Suficiente"?** (85 también es ≥ 70 y ≥ 60)

Porque Python revisa **de arriba hacia abajo** y **se detiene en la primera condición verdadera**:
- ¿85 >= 90? No → sigue.
- ¿85 >= 80? **Sí** → imprime "Muy bien" y **se salta todo lo demás**.

⚠️ **Por eso el orden importa.** Mira qué pasa si lo pones al revés:

````python
nota = 95

# ❌ Orden incorrecto: la condición más "fácil" va primero y atrapa todo
if nota >= 60:
    print("Suficiente")
elif nota >= 90:
    print("Excelente")   # ¡nunca llega aquí, aunque la nota sea 95!
````

### 4.4 `elif` vs. varios `if` seguidos (diferencia clave)

- Varios **`if`** separados → cada uno se revisa **siempre**, de forma independiente. Pueden ejecutarse **varios**.
- **`if` / `elif`** → están encadenados: se ejecuta **solo uno**.

````python
n = 12

print("--- Con if separados (se revisan todos) ---")
if n % 2 == 0:
    print("Es divisible entre 2")
if n % 3 == 0:
    print("Es divisible entre 3")
if n % 4 == 0:
    print("Es divisible entre 4")

print("--- Con if/elif (solo el primero que se cumpla) ---")
if n % 2 == 0:
    print("Es divisible entre 2")
elif n % 3 == 0:
    print("Es divisible entre 3")
elif n % 4 == 0:
    print("Es divisible entre 4")
````

> 🧭 **Regla:** ¿Las opciones se **excluyen** entre sí (solo puede pasar una)? → `if/elif/else`.
> ¿Son preguntas **independientes** (pueden cumplirse varias)? → varios `if`.

### 4.5 `if` dentro de `if` (anidados)

````python
edad = 20
tiene_licencia = False

if edad >= 18:
    if tiene_licencia:
        print("Puedes conducir.")
    else:
        print("Tienes edad, pero te falta la licencia.")
else:
    print("Todavía eres menor de edad.")
````

## 5. Repetición con `for`

### La idea
`for` significa **"por cada"**. Úsalo cuando:
- **sabes cuántas veces** quieres repetir algo, o
- tienes una **colección** (lista, texto, rango de números) y quieres hacer algo **con cada elemento**.

```python
for variable in colección:
    # esto se repite una vez por cada elemento
```

En cada vuelta, `variable` toma el siguiente valor de la colección.

### 5.1 `range()`: generar números

| Escribes | Obtienes | Explicación |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | Del 0 hasta 5 **sin incluir el 5** |
| `range(1, 6)` | 1, 2, 3, 4, 5 | Desde 1 hasta 6 sin incluir el 6 |
| `range(0, 11, 2)` | 0, 2, 4, 6, 8, 10 | De 2 en 2 |
| `range(10, 0, -1)` | 10, 9, …, 1 | Hacia atrás |

⚠️ El número final **nunca se incluye**. Para contar del 1 al 10 usa `range(1, 11)`.

````python
for i in range(5):
    print("Vuelta número", i)
````

````python
# Contar del 1 al 10
for numero in range(1, 11):
    print(numero)
````

````python
# Cuenta regresiva
for n in range(10, 0, -1):
    print(n)
print("¡Despegue! 🚀")
````

### 5.2 Recorrer una lista o un texto

````python
frutas = ["manzana", "plátano", "uva"]

for fruta in frutas:
    print(f"Me gusta la {fruta}")
````

````python
for letra in "Python":
    print(letra)
````

### 5.3 El patrón más importante: el **acumulador**

Muchísimos problemas se resuelven así:
1. Creas una variable **antes** del `for` (por ejemplo `total = 0`).
2. **Dentro** del `for` la vas actualizando.
3. **Después** del `for` usas el resultado.

````python
# Sumar los números del 1 al 5
total = 0                     # 1. empieza en 0
for n in range(1, 6):
    total = total + n         # 2. se va acumulando
    print(f"Sumé {n}, el total va en {total}")
print("Resultado final:", total)   # 3. se usa al terminar (fuera del for)
````

Sigue la ejecución "a mano" (esto es lo que hace la computadora):

| Vuelta | `n` | `total` antes | `total` después |
|---|---|---|---|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |
| 5 | 5 | 10 | 15 |

> 💡 Hacer esta tabla a mano es **la mejor forma** de entender cualquier ciclo.

Variante: el **contador** (cuántos cumplen algo). Aquí combinamos `for` + `if`:

````python
numeros = [4, 7, 10, 13, 16, 21, 22]
pares = 0

for n in numeros:
    if n % 2 == 0:
        pares += 1

print(f"Hay {pares} números pares en la lista")
````

### 5.4 `for` dentro de `for` (anidados)
El `for` de adentro se ejecuta **completo** en cada vuelta del de afuera.

````python
for fila in range(1, 4):
    for columna in range(1, 4):
        print(f"({fila},{columna})", end="  ")   # end="  " evita el salto de línea
    print()   # salto de línea al terminar cada fila
````

## 6. Repetición con `while`

### La idea
`while` significa **"mientras"**. Repite **mientras una condición sea verdadera**. Úsalo cuando **NO sabes cuántas veces** tendrás que repetir: depende de algo que va cambiando.

```python
while condición:
    # se repite mientras la condición sea True
    # ¡algo aquí adentro debe cambiar la condición!
```

**Cómo funciona cada vuelta:**
1. Python revisa la condición.
2. Si es `True` → ejecuta el bloque y **vuelve al paso 1**.
3. Si es `False` → sale del ciclo y sigue con el resto del programa.

### 6.1 Ejemplo: ¿cuántos meses para ahorrar?
No sabemos de antemano cuántos meses serán, así que `while` es la herramienta natural.

````python
meta = 1000
ahorro = 0
ahorro_mensual = 150
meses = 0

while ahorro < meta:           # mientras NO haya llegado a la meta...
    ahorro += ahorro_mensual   # ...ahorro un mes más
    meses += 1
    print(f"Mes {meses}: llevo ${ahorro}")

print(f"Llegué a la meta en {meses} meses.")
````

### 6.2 ⚠️ Cuidado con el ciclo infinito

Si **nada dentro del `while` cambia la condición**, nunca termina. Por ejemplo:

```python
x = 1
while x > 0:
    print(x)      # x nunca cambia → se repite para siempre
```

Si te pasa: **Kernel → Interrupt** (o el botón ⏹ de la barra).

**Checklist de todo `while`:**
1. ¿La variable de la condición tiene un valor **antes** del ciclo?
2. ¿Algo **dentro** del ciclo la modifica?
3. ¿Esa modificación **acerca** la condición a ser `False`?

### 6.3 `while` para validar datos (repetir hasta que la respuesta sea correcta)
Este es un caso clásico de `while`: no sabes cuántas veces se equivocará el usuario.

````python
edad = -1   # valor inicial inválido para que entre al ciclo

while edad < 0 or edad > 120:
    edad = int(input("Escribe tu edad (0 a 120): "))
    if edad < 0 or edad > 120:
        print("Eso no es una edad válida, intenta otra vez.")

print(f"Perfecto, tienes {edad} años.")
````

### 6.4 Juego: adivina el número 🎲

````python
import random

secreto = random.randint(1, 20)   # número al azar entre 1 y 20
intento = 0
intentos = 0

while intento != secreto:
    intento = int(input("Adivina el número (1 a 20): "))
    intentos += 1
    if intento < secreto:
        print("Más alto ⬆️")
    elif intento > secreto:
        print("Más bajo ⬇️")

print(f"¡Correcto! Era {secreto}. Lo lograste en {intentos} intentos.")
````

> Observa cómo este programa usa **los 3 ingredientes**: secuencia (las líneas en orden), repetición (`while`, porque no sabemos cuántos intentos harán falta) y decisión (`if/elif`, para dar la pista correcta).

### 6.5 Todo `for` se puede escribir como `while` (pero es más largo)

````python
# Con for
for i in range(1, 6):
    print(i)

print("---")

# Lo mismo con while: tienes que manejar tú el contador
i = 1
while i <= 5:
    print(i)
    i += 1     # si olvidas esta línea → ciclo infinito
````

Por eso: **si sabes cuántas veces, usa `for`** (es más corto y no se puede volver infinito).

## 7. 🧭 ¿Cuándo uso cada uno? La guía para decidir

Cuando leas un problema, hazte estas preguntas **en este orden**:

```
¿Tengo que REPETIR algo?
│
├── NO → ¿Depende de una condición?
│        ├── NO  → solo escribe las líneas (secuencia)
│        └── SÍ  → ¿Cuántos caminos posibles hay?
│                  ├── 1 camino ("haz esto solo si...")     → if
│                  ├── 2 caminos ("esto o aquello")         → if / else
│                  └── 3 o más ("elige una de varias")      → if / elif / else
│
└── SÍ → ¿Sé CUÁNTAS veces, o tengo una lista/rango que recorrer?
         ├── SÍ → for
         └── NO, repito HASTA QUE pase algo              → while
```

### Palabras clave en el enunciado

| Si el problema dice… | Probablemente necesitas |
|---|---|
| "si…", "en caso de que…", "solo cuando…" | `if` |
| "si no…", "de lo contrario…" | `else` |
| "si es A…, si es B…, si es C…" (categorías) | `if / elif / else` |
| "para cada…", "por cada…", "todos los…", "del 1 al N", "N veces" | `for` |
| "recorre la lista…", "cada letra…", "cada elemento…" | `for` |
| "mientras…", "hasta que…", "repite hasta…", "cuántos intentos/meses/pasos hacen falta" | `while` |
| "pide un dato hasta que sea válido" | `while` |
| "el primero que cumpla…" | `for` + `break` (o `while`) |

### Ejemplos rápidos: ¿qué usarías?

| Problema | Respuesta | Por qué |
|---|---|---|
| Imprimir la tabla del 7 (del 1 al 10) | `for` | Sabes que son 10 veces |
| Decir si un número es positivo, negativo o cero | `if/elif/else` | 3 caminos, solo uno ocurre |
| ¿Cuántos años tarda en duplicarse una inversión al 5%? | `while` | No sabes cuántos años: repites hasta que se duplique |
| Calcular el promedio de una lista de notas | `for` | Recorres cada nota |
| Pedir una contraseña hasta que sea correcta | `while` | No sabes cuántos intentos |
| Cobrar 10% de descuento si la compra supera $500 | `if` | Un solo caso especial |
| Contar cuántas vocales tiene una palabra | `for` + `if` | Recorres cada letra y preguntas si es vocal |

## 8. `break` y `continue`: controlar un ciclo desde adentro

- **`break`** → sale del ciclo **inmediatamente** (lo termina).
- **`continue`** → se salta el **resto de esta vuelta** y pasa a la siguiente.

````python
# break: busca el primer múltiplo de 7 mayor que 50
for n in range(51, 100):
    if n % 7 == 0:
        print("El primero es", n)
        break          # ya lo encontré, no tiene sentido seguir buscando
````

````python
# continue: imprime del 1 al 10 saltándote los múltiplos de 3
for n in range(1, 11):
    if n % 3 == 0:
        continue       # salta a la siguiente vuelta sin imprimir
    print(n)
````

````python
# while True + break: un patrón muy común para menús
while True:
    opcion = input("Escribe 'salir' para terminar, o cualquier otra cosa para seguir: ")
    if opcion == "salir":
        break
    print(f"Escribiste: {opcion}")
print("Programa terminado.")
````

## 9. Listas: guardar varios datos en una sola variable

Una lista va entre corchetes `[ ]` y sus elementos se separan con comas. Cada elemento tiene una **posición (índice)** que **empieza en 0**.

````python
notas = [8, 6, 9, 10, 7]

print(notas[0])       # primer elemento → 8
print(notas[2])       # tercer elemento → 9
print(notas[-1])      # el último → 7
print(len(notas))     # cuántos elementos hay → 5

notas.append(5)       # agregar al final
print(notas)

notas[1] = 7          # cambiar un elemento
print(notas)
````

````python
# Funciones útiles que ya vienen en Python
numeros = [4, 9, 1, 7, 3]
print("Suma:", sum(numeros))
print("Máximo:", max(numeros))
print("Mínimo:", min(numeros))
print("Ordenada:", sorted(numeros))
print("¿Está el 7?", 7 in numeros)
````

````python
# Construir una lista con un for: los cuadrados del 1 al 10
cuadrados = []
for n in range(1, 11):
    cuadrados.append(n ** 2)
print(cuadrados)
````

## 10. Funciones: darle nombre a un bloque de código para reutilizarlo

```python
def nombre_de_la_funcion(parametros):
    # instrucciones
    return resultado
```

- `def` define la función (todavía no la ejecuta).
- Los **parámetros** son los datos que recibe.
- `return` **devuelve** el resultado a quien la llamó.

````python
def area_rectangulo(base, altura):
    return base * altura

print(area_rectangulo(5, 3))
print(area_rectangulo(10, 2.5))

resultado = area_rectangulo(4, 4)
print(f"El área es {resultado}")
````

## 11. Cómo resolver un problema paso a paso

**Nunca empieces escribiendo código.** Sigue estos pasos:

1. **Entiende** el problema: ¿qué datos tengo (entrada)? ¿qué debo obtener (salida)?
2. **Resuélvelo a mano** con un ejemplo pequeño.
3. **Escribe los pasos en español** (pseudocódigo).
4. **Identifica los ingredientes:** ¿hay decisiones (`if`)? ¿repeticiones (`for`/`while`)?
5. **Traduce** a Python línea por línea.
6. **Prueba** con varios valores, incluidos casos raros (0, negativos, listas vacías).

### Ejemplo: "Dada una lista de temperaturas, di cuántos días hizo calor (> 30°), cuál fue la máxima y el promedio."

**Paso 1:** Entrada → una lista de números. Salida → 3 datos: cantidad de días calurosos, máxima, promedio.

**Paso 2:** Con `[28, 32, 35, 29]`: días calurosos = 2 (32 y 35); máxima = 35; promedio = 124 / 4 = 31.

**Paso 3 (pseudocódigo):**
```
contador_calor = 0, suma = 0, maxima = la primera temperatura
por cada temperatura de la lista:
    sumarla a suma
    si es mayor que 30: sumar 1 a contador_calor
    si es mayor que maxima: maxima = esta temperatura
promedio = suma / cantidad de temperaturas
mostrar resultados
```

**Paso 4:** "por cada temperatura" → `for`. "si es mayor que…" → `if` (dos `if` **independientes**, no `elif`, porque las dos cosas pueden pasar a la vez).

**Paso 5:**

````python
temperaturas = [28, 32, 35, 29, 31, 27, 33]

contador_calor = 0
suma = 0
maxima = temperaturas[0]

for t in temperaturas:
    suma += t
    if t > 30:
        contador_calor += 1
    if t > maxima:
        maxima = t

promedio = suma / len(temperaturas)

print(f"Días de calor: {contador_calor}")
print(f"Temperatura máxima: {maxima}")
print(f"Promedio: {promedio:.1f}")
````

### Ejercicio 4: Ecuación cuadrática ax² + bx + c = 0

**Fórmula general:**  x = (−b ± √(b² − 4ac)) / 2a

**Cómo pensarlo:** primero calculamos el **discriminante** d = b² − 4ac. Según su valor hay **3 casos excluyentes**:
- d > 0 → dos soluciones reales distintas
- d = 0 → una solución (doble)
- d < 0 → no hay soluciones reales

3 casos, solo uno ocurre → `if / elif / else`.

````python
import math

a = 1
b = -5
c = 6     # x² - 5x + 6 = 0  → soluciones 2 y 3

d = b ** 2 - 4 * a * c
print(f"Discriminante: {d}")

if d > 0:
    x1 = (-b + math.sqrt(d)) / (2 * a)
    x2 = (-b - math.sqrt(d)) / (2 * a)
    print(f"Dos soluciones: x1 = {x1}, x2 = {x2}")
elif d == 0:
    x = -b / (2 * a)
    print(f"Una solución (doble): x = {x}")
else:
    print("No tiene soluciones reales")
````

Prueba con `a=1, b=2, c=1` (d = 0) y con `a=1, b=0, c=4` (d < 0).

### Ejercicio 9: ¿Es un número primo?
Un número primo solo es divisible entre 1 y él mismo (2, 3, 5, 7, 11, 13…).

**Cómo pensarlo:** pruebo dividir entre 2, 3, 4… hasta n−1. Si **alguno** divide exacto (`n % d == 0`), no es primo y **ya no hace falta seguir** → `for` + `if` + `break`.

````python
n = 29

es_primo = True          # suponemos que sí, hasta demostrar lo contrario
if n < 2:
    es_primo = False
else:
    for d in range(2, n):
        if n % d == 0:
            es_primo = False
            print(f"{n} es divisible entre {d}")
            break        # ya sabemos que no es primo

if es_primo:
    print(f"{n} es primo")
else:
    print(f"{n} NO es primo")
````

Ahora lo convertimos en **función** para reutilizarlo y listar todos los primos hasta 100.
(Truco matemático: basta con probar divisores hasta √n, así es mucho más rápido.)

````python
import math

def es_primo(n):
    if n < 2:
        return False
    for d in range(2, int(math.sqrt(n)) + 1):
        if n % d == 0:
            return False     # return también termina la función (como un break)
    return True

primos = []
for numero in range(1, 101):
    if es_primo(numero):
        primos.append(numero)

print(primos)
print(f"Hay {len(primos)} primos entre 1 y 100")
````

### Ejercicio 10: Máximo común divisor (algoritmo de Euclides)
**Idea de Euclides:** el MCD de `a` y `b` es igual al MCD de `b` y `a % b`. Se repite **hasta que** el residuo sea 0.

**Cómo pensarlo:** "hasta que el residuo sea 0" → no sé cuántos pasos → `while`.

````python
a = 48
b = 18
a_original, b_original = a, b

while b != 0:
    print(f"a = {a}, b = {b}, residuo = {a % b}")
    a, b = b, a % b      # asignación doble: a toma el valor de b, b toma el residuo

print(f"MCD({a_original}, {b_original}) = {a}")

# El mínimo común múltiplo sale del MCD:
mcm = a_original * b_original // a
print(f"mcm({a_original}, {b_original}) = {mcm}")
````

### Ejercicio 13: Sumar los dígitos de un número
**Problema:** 4827 → 4 + 8 + 2 + 7 = 21.

**Cómo pensarlo (truco matemático):**
- `n % 10` da el **último dígito** (4827 % 10 = 7).
- `n // 10` **quita** el último dígito (4827 // 10 = 482).
- Repito **hasta que** n sea 0 → `while`.

````python
n = 4827
original = n
suma_digitos = 0

while n > 0:
    digito = n % 10
    suma_digitos += digito
    print(f"Tomo el dígito {digito}, queda {n // 10}")
    n = n // 10

print(f"La suma de los dígitos de {original} es {suma_digitos}")
````

### Ejercicio 15: Conjetura de Collatz
Empieza con cualquier número positivo:
- si es **par**, divídelo entre 2;
- si es **impar**, multiplícalo por 3 y súmale 1.

Se cree que **siempre** se llega a 1, pero nadie lo ha demostrado. ¿Cuántos pasos tarda?

**Cómo pensarlo:** "hasta llegar a 1" → `while`. "Si es par… si no…" → `if / else`.

````python
n = 27
pasos = 0
secuencia = [n]

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    pasos += 1
    secuencia.append(n)

print(f"Llegó a 1 en {pasos} pasos")
print(f"El valor más alto alcanzado fue {max(secuencia)}")
````

## 14. Errores comunes y cómo leerlos

Los errores **no son malos**: Python te está diciendo exactamente qué falló. Lee **la última línea** del mensaje rojo: ahí está el tipo de error y la pista.

| Error | Qué significa | Causa típica |
|---|---|---|
| `SyntaxError` | Escribiste algo que Python no entiende | Faltan los `:` después de `if`/`for`/`while`/`def`, paréntesis o comillas sin cerrar, usar `=` en lugar de `==` |
| `IndentationError` | La indentación está mal | Mezclar espacios y tabs, o olvidar indentar lo que va dentro de un `if`/`for` |
| `NameError: name 'x' is not defined` | Usas una variable que no existe | Error de escritura (`Nombre` vs `nombre`) o no ejecutaste la celda donde se creó |
| `TypeError` | Operación con tipos incompatibles | Sumar texto y número: `"Edad: " + 25` (usa f-strings o `str()`) |
| `ValueError` | El valor no sirve para esa operación | `int("hola")` o `int("3.5")` |
| `ZeroDivisionError` | Dividiste entre cero | Promedio de una lista vacía |
| `IndexError: list index out of range` | Pediste una posición que no existe | `lista[5]` en una lista de 5 elementos (las posiciones van de 0 a 4) |

Ejecuta esta celda para ver cómo luce un error (es a propósito):

````python
edad = 25
print("Tengo " + edad + " años")   # TypeError: no se puede unir texto con número
````

**La corrección:**

````python
edad = 25
print(f"Tengo {edad} años")
````
