## 1. La idea: el programa elige un camino

Piensa en cómo decides en la vida real:
> *"**Si** llueve, llevo paraguas. **Si no**, llevo lentes de sol."*

Eso es exactamente un `if` / `else`. La computadora:
1. **Hace una pregunta** que se responde con `True` o `False` (como en la Clase 5).
2. **Si la respuesta es `True`**, ejecuta un bloque de código.
3. **Si es `False`**, lo salta (o ejecuta otro bloque, si hay `else`).

| Palabra | Significa | Se usa para |
|---|---|---|
| `if` | "si..." | La primera pregunta |
| `elif` | "si no, entonces si..." | Preguntas extra (solo se revisan si las anteriores fueron `False`) |
| `else` | "si no..." / "en cualquier otro caso" | El plan B: cuando ninguna pregunta fue `True` |

---
## 2. `if` solo: "haz esto solo si..."

```python
if condición:
    lo que se hace si la condición es True
```

Reglas:
- Después de la condición van **dos puntos `:`**.
- Lo que pertenece al `if` va **movido a la derecha** (4 espacios o Tab).

````python
temperatura = 35

if temperatura > 30:
    print("Hace calor, toma agua.")

print("Este mensaje sale siempre.")
````

Cómo lo lee Python:
1. ¿`35 > 30`? → **True** → ejecuta el `print` de adentro.
2. El último `print` **no está movido a la derecha**, así que está **fuera** del `if`: se ejecuta siempre.

👉 **Cambia `temperatura` a `20` y vuelve a ejecutar.** El primer mensaje desaparece, el segundo no.

---
## 3. La sangría (indentación): qué está dentro y qué está fuera

En Python, los espacios al inicio de la línea **no son decoración**: le dicen a Python **qué líneas pertenecen al `if`**.

Jupyter la pone solo: cuando escribes `:` y pulsas **Enter**, la siguiente línea ya empieza movida. Para "salirte" del `if`, borra esos espacios.

````python
dinero = 50

if dinero >= 100:
    print("Puedes comprar el juego")     # dentro del if
    print("¡Disfrútalo!")                # también dentro del if

print("Gracias por visitar la tienda")  # fuera del if: sale siempre
````

Con `dinero = 50`, la condición es `False`, así que Python **salta las dos líneas de adentro** y solo muestra el mensaje de afuera.

👉 Cambia `dinero` a `150` y verás las tres líneas.

## 4. `if` + `else`: dos caminos

```python
if condición:
    camino A (si es True)
else:
    camino B (si es False)
```

Siempre se ejecuta **uno de los dos**, nunca ambos y nunca ninguno. El `else` **no lleva condición**: es "todo lo demás".

````python
numero = 7

if numero % 2 == 0:
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
````

¿Te acuerdas del ejercicio 0.3, donde escribías "es par" a mano? Ahora **la computadora decide sola**. Cambia `numero` por cualquier valor y siempre acierta.

````python
edad = int(input("¿Cuántos años tienes? "))

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
````

---
## 5. `if` + `elif` + `else`: tres o más caminos

Cuando hay **varias opciones** y **solo una** puede ocurrir:

```python
if condición_1:
    camino 1
elif condición_2:
    camino 2
elif condición_3:
    camino 3
else:
    camino por defecto
```

Puedes poner **todos los `elif` que necesites**. El `else` del final es opcional.

````python
nota = 85

if nota >= 90:
    print("Excelente")
elif nota >= 80:
    print("Muy bien")
elif nota >= 70:
    print("Bien")
elif nota >= 60:
    print("Suficiente")
else:
    print("Reprobado")
````

**Paso a paso con `nota = 85`:**
1. ¿`85 >= 90`? → False → sigue al siguiente.
2. ¿`85 >= 80`? → **True** → muestra "Muy bien" y **se salta todo lo demás**.

⚠️ **La regla más importante:** Python revisa **de arriba hacia abajo** y **se detiene en la primera condición `True`**. Aunque 85 también sea mayor que 70 y que 60, esas ya no se revisan. **Solo se ejecuta UN camino.**

👉 Prueba con `nota = 95`, `72`, `60` y `40`.

---
## 6. ¿Por qué el orden importa?

Como Python se detiene en la primera condición verdadera, si pones primero la condición más "fácil", atrapa todo:

````python
nota = 95

# ❌ Orden incorrecto: la condición más fácil va primero
if nota >= 60:
    print("Suficiente")
elif nota >= 90:
    print("Excelente")     # ¡nunca llega aquí, aunque la nota sea 95!
````

Con 95, la primera pregunta (`95 >= 60`) ya es `True`, así que dice "Suficiente" y nunca llega a "Excelente".

> 💡 **Regla:** cuando uses `>=` con rangos, pon primero la condición **más exigente** (el número más alto).

## 7. `elif` vs. varios `if`

- **`if` / `elif`** → están encadenados: se ejecuta **solo uno**.
- **Varios `if` separados** → cada uno se revisa **siempre**, por su cuenta: pueden ejecutarse **varios**.

````python
n = 12

print("--- Con if separados (se revisan todos) ---")
if n % 2 == 0:
    print("Es divisible entre 2")
if n % 3 == 0:
    print("Es divisible entre 3")
if n % 4 == 0:
    print("Es divisible entre 4")

print("--- Con if / elif (solo el primero que se cumpla) ---")
if n % 2 == 0:
    print("Es divisible entre 2")
elif n % 3 == 0:
    print("Es divisible entre 3")
elif n % 4 == 0:
    print("Es divisible entre 4")
````

> 🧭 **¿Cuál uso?**
> - ¿Las opciones **se excluyen** entre sí (solo puede pasar una)? → `if` / `elif` / `else`. *Ejemplo: una calificación es "Excelente" O "Bien", no las dos.*
> - ¿Son preguntas **independientes** (pueden cumplirse varias)? → varios `if`. *Ejemplo: un número puede ser divisible entre 2 Y entre 3.*

## 9. `if` dentro de `if` (anidados)

A veces una decisión depende de otra. Fíjate que cada nivel va **más a la derecha**:

````python
edad = 20
tiene_licencia = False

if edad >= 18:
    if tiene_licencia:
        print("Puedes conducir.")
    else:
        print("Tienes edad, pero te falta la licencia.")
else:
    print("Todavía eres menor de edad.")
````

Primero pregunta la edad. **Solo si** es mayor de edad, hace la segunda pregunta (la licencia).

## 10. Errores comunes

| Error | Ejemplo | Qué pasa | Corrección |
|---|---|---|---|
| Olvidar los `:` | `if edad >= 18` | `SyntaxError` | `if edad >= 18:` |
| Usar `=` en vez de `==` | `if edad = 18:` | `SyntaxError` | `if edad == 18:` |
| Olvidar la sangría | `if x > 5:` y en la siguiente línea `print("hola")` sin espacios | `IndentationError` | Mover el `print` 4 espacios a la derecha |
| Ponerle condición al `else` | `else edad < 18:` | `SyntaxError` | `else:` (el `else` nunca lleva condición; si necesitas una, usa `elif`) |
| Orden incorrecto en rangos | `if nota >= 60:` antes que `>= 90` | No da error, pero **da un resultado equivocado** | Poner primero la más exigente |

Ejecuta esta celda para ver un error (es a propósito): le falta la sangría al `print`.

````python
edad = 20
if edad >= 18:
print("Eres mayor de edad")
````

## 11. Guía rápida: ¿cuál uso?

| Si el problema dice... | Usa | Ejemplo |
|---|---|---|
| "haz esto **solo si**..." (1 camino) | `if` | Si la compra pasa de $1000, regala envío |
| "si... **si no**..." (2 caminos) | `if` / `else` | Par o impar; mayor o menor de edad |
| "si es A..., si es B..., si es C..." (3 o más, solo uno ocurre) | `if` / `elif` / `else` | Calificaciones; positivo, negativo o cero |
| Varias preguntas que **pueden cumplirse a la vez** | varios `if` | Divisible entre 2, entre 3, entre 5... |

**Cómo pensar un problema de decisiones:**
1. ¿Cuántos **caminos** posibles hay?
2. ¿Qué **pregunta** (comparación) decide cada camino?
3. Si hay rangos, ¿en qué **orden** los pongo?
4. ¿Hay un caso "por defecto" para el `else`?

# 📘 Clase 7: Repetir con `for`

Hasta ahora, si querías hacer algo 10 veces, tenías que escribirlo 10 veces. Con `for` lo escribes **una sola vez** y Python lo repite.

`for` significa **"por cada"**. Úsalo cuando:
- **sabes cuántas veces** quieres repetir algo, o
- tienes una **colección** (una lista, un texto, un rango de números) y quieres hacer algo **con cada elemento**.

```python
for variable in colección:
    lo que se repite
```

En cada vuelta, `variable` toma **el siguiente valor** de la colección. Igual que en el `if`: **dos puntos `:`** y **sangría**.

## 7.1 `range()`: generar números

| Escribes | Obtienes | Explicación |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | Del 0 hasta 5, **sin incluir el 5** |
| `range(1, 6)` | 1, 2, 3, 4, 5 | Desde 1 hasta 6, sin incluir el 6 |
| `range(0, 11, 2)` | 0, 2, 4, 6, 8, 10 | De 2 en 2 (el tercer número es el "salto") |
| `range(10, 0, -1)` | 10, 9, ..., 1 | Hacia atrás |

⚠️ **El número final nunca se incluye.** Para contar del 1 al 10 usa `range(1, 11)`.

````python
for i in range(5):
    print("Vuelta número", i)
````

````python
# Contar del 1 al 10
for numero in range(1, 11):
    print(numero)
````

````python
# De 2 en 2
for n in range(0, 11, 2):
    print(n)
````

````python
# Cuenta regresiva
for n in range(10, 0, -1):
    print(n)
print("¡Despegue! 🚀")      # fuera del for: se ejecuta una sola vez, al final
````

> 💡 Fíjate en el último `print`: **no tiene sangría**, así que está **fuera** del `for` y se ejecuta **una vez**, cuando termina el ciclo.

## 7.2 Recorrer un texto
Un texto es una colección de letras. `for` las recorre una por una:

````python
for letra in "Python":
    print(letra)
````

## 7.3 El patrón más importante: el **acumulador**

Muchísimos problemas se resuelven así:
1. Creas una variable **antes** del `for` (por ejemplo `total = 0`).
2. **Dentro** del `for` la vas actualizando.
3. **Después** del `for` usas el resultado.

````python
# Sumar los números del 1 al 5
total = 0                      # 1. empieza en 0
for n in range(1, 6):
    total = total + n          # 2. se va acumulando
    print(f"Sumé {n}, el total va en {total}")
print("Resultado final:", total)   # 3. se usa al terminar
````

Así lo ejecuta la computadora (haz esta tabla a mano cuando no entiendas un ciclo):

| Vuelta | `n` | `total` antes | `total` después |
|---|---|---|---|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |
| 5 | 5 | 10 | 15 |

> ⚠️ Si vas a **multiplicar** (por ejemplo un factorial), el acumulador empieza en **1**, no en 0. Si empieza en 0, todo da 0.

## 7.4 El **contador**: `for` + `if`
Para contar cuántos elementos cumplen algo, combinas `for` con `if`:

````python
contador_pares = 0
for n in range(1, 21):
    if n % 2 == 0:
        contador_pares += 1
print(f"Del 1 al 20 hay {contador_pares} números pares")
````

## 7.5 `for` dentro de `for` (anidados)
El `for` de adentro se ejecuta **completo** en cada vuelta del de afuera:

````python
for fila in range(1, 4):
    for columna in range(1, 4):
        print(f"({fila},{columna})", end="  ")   # end="  " evita el salto de línea
    print()                                       # salto de línea al terminar cada fila
````

````python
# Tablas de multiplicar del 1 al 3
for tabla in range(1, 4):
    print(f"--- Tabla del {tabla} ---")
    for i in range(1, 6):
        print(f"{tabla} x {i} = {tabla * i}")
````

# 📘 Clase 8: Repetir con `while`

`while` significa **"mientras"**. Repite **mientras una condición sea verdadera**. Úsalo cuando **NO sabes cuántas veces** vas a repetir, porque depende de algo que va cambiando.

```python
while condición:
    lo que se repite
    # ¡algo aquí adentro debe cambiar la condición!
```

**Cómo funciona cada vuelta:**
1. Python revisa la condición.
2. Si es `True` → ejecuta el bloque y **vuelve al paso 1**.
3. Si es `False` → sale del ciclo y sigue con el resto del programa.

Es como un `if` que **se repite** mientras la respuesta siga siendo `True`.

## 8.1 Ejemplo: ¿cuántos meses para juntar mi meta?

````python
meta = 1000
ahorro = 0
ahorro_mensual = 150
meses = 0

while ahorro < meta:           # mientras NO haya llegado a la meta...
    ahorro += ahorro_mensual   # ...ahorro un mes más
    meses += 1
    print(f"Mes {meses}: llevo ${ahorro}")

print(f"Llegué a la meta en {meses} meses.")
````

Nunca le dijimos cuántas veces repetir: **el ciclo lo descubrió solo**. Esa es la diferencia con `for`.

## 8.2 ⚠️ Cuidado con el ciclo infinito

Si **nada dentro del `while` cambia la condición**, nunca termina:

```python
x = 1
while x > 0:
    print(x)      # x nunca cambia → se repite para siempre
```

**Si te pasa en Jupyter:** menú **Kernel → Interrupt** (o el botón ⏹ de la barra).

**Checklist para todo `while`:**
1. ¿La variable de la condición tiene un valor **antes** del ciclo?
2. ¿Algo **dentro** del ciclo la modifica?
3. ¿Esa modificación **acerca** la condición a ser `False`?

## 8.3 Todo `for` se puede escribir con `while`... pero es más largo

````python
# Con for
for i in range(1, 6):
    print(i)

print("---")

# Lo mismo con while: tienes que manejar tú el contador
i = 1
while i <= 5:
    print(i)
    i += 1     # si olvidas esta línea → ciclo infinito
````

Por eso: **si sabes cuántas veces, usa `for`**. Es más corto y no se puede volver infinito.

## 8.4 `while` para validar datos
Un uso clásico: repetir la pregunta **hasta que** la respuesta sea válida. No sabes cuántas veces se va a equivocar el usuario.

````python
edad = -1   # valor inicial inválido, para que entre al ciclo

while edad < 0 or edad > 120:
    edad = int(input("Escribe tu edad (0 a 120): "))
    if edad < 0 or edad > 120:
        print("Esa no es una edad válida, intenta otra vez.")

print(f"Perfecto, tienes {edad} años.")
````

## 8.5 `break` y `continue`: controlar un ciclo desde adentro

- **`break`** → sale del ciclo **inmediatamente**.
- **`continue`** → se salta **el resto de esta vuelta** y pasa a la siguiente.

Funcionan en `for` y en `while`.

````python
# break: busca el primer múltiplo de 7 mayor que 50
for n in range(51, 100):
    if n % 7 == 0:
        print("El primero es", n)
        break          # ya lo encontré, no tiene sentido seguir
````

````python
# continue: del 1 al 10, saltándose los múltiplos de 3
for n in range(1, 11):
    if n % 3 == 0:
        continue       # salta a la siguiente vuelta sin imprimir
    print(n)
````

## 8.6 `while True` + `break`: el patrón del menú
`while True` es un ciclo que **nunca termina por sí solo**. Se usa cuando la salida se decide **adentro**, con un `break`:

````python
while True:
    opcion = input("Escribe 'salir' para terminar, o cualquier otra cosa para seguir: ")
    if opcion == "salir":
        break
    print(f"Escribiste: {opcion}")
print("Programa terminado.")
````

## 8.7 Juego: adivina el número 🎲

````python
import random

secreto = random.randint(1, 20)   # número al azar entre 1 y 20
intento = 0
intentos = 0

while intento != secreto:
    intento = int(input("Adivina el número (1 a 20): "))
    intentos += 1
    if intento < secreto:
        print("Más alto ⬆️")
    elif intento > secreto:
        print("Más bajo ⬇️")

print(f"¡Correcto! Era {secreto}. Lo lograste en {intentos} intentos.")
````

Este programa usa **los 3 ingredientes de la programación**: secuencia (líneas en orden), **repetición** (`while`, porque no sabemos cuántos intentos hacen falta) y **decisión** (`if` / `elif`, para dar la pista).

# 📘 Clase 9: ¿`for` o `while`? Cómo decidir

Esta es la pregunta que más confunde al empezar. Hazte esto **en orden**:

```
¿Tengo que REPETIR algo?
│
├── NO → ¿Depende de una condición?
│        ├── NO  → solo escribe las líneas (secuencia)
│        └── SÍ  → ¿Cuántos caminos hay?
│                  ├── 1 camino    → if
│                  ├── 2 caminos   → if / else
│                  └── 3 o más     → if / elif / else
│
└── SÍ → ¿Sé CUÁNTAS veces, o tengo una lista/rango que recorrer?
         ├── SÍ → for
         └── NO, repito HASTA QUE pase algo → while
```

## Palabras clave en el enunciado

| Si el problema dice... | Probablemente necesitas |
|---|---|
| "si...", "solo cuando..." | `if` |
| "si no...", "de lo contrario..." | `else` |
| "si es A..., si es B..., si es C..." | `if` / `elif` / `else` |
| "para cada...", "todos los...", "del 1 al N", "N veces" | `for` |
| "recorre la lista...", "cada letra...", "cada elemento..." | `for` |
| "mientras...", "hasta que...", "¿cuántos intentos / meses / años hacen falta?" | `while` |
| "pide un dato hasta que sea válido" | `while` |
| "el primero que cumpla..." | `for` + `break` (o `while`) |

## El mismo problema, dos preguntas distintas: Fibonacci
La sucesión de Fibonacci: cada número es la suma de los dos anteriores (0, 1, 1, 2, 3, 5, 8, 13, ...).

````python
# "Los primeros 15 términos" → sé cuántos → for
a, b = 0, 1
for i in range(15):
    print(a, end=" ")
    a, b = b, a + b
````

````python
# "Todos los términos menores que 1000" → no sé cuántos serán → while
a, b = 0, 1
while a < 1000:
    print(a, end=" ")
    a, b = b, a + b
````

> 💡 **Mismo tema, distinta pregunta, distinto ciclo.** Lo que decide no es el tema, sino **la pregunta**.
>
> (`a, b = b, a + b` es una **asignación doble**: `a` toma el valor de `b` y `b` toma `a + b`, las dos al mismo tiempo.)

# 📘 Clase 10: Listas

Una **lista** guarda **varios datos en una sola variable**. Va entre corchetes `[ ]` y los elementos se separan con comas.

```python
frutas = ["manzana", "plátano", "uva"]
notas = [8, 6, 9, 10, 7]
```

Cada elemento tiene una **posición (índice)** que **empieza en 0**, no en 1:

| Índice | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| `notas` | 8 | 6 | 9 | 10 | 7 |
| Desde el final | -5 | -4 | -3 | -2 | -1 |

## 10.1 Leer elementos

````python
notas = [8, 6, 9, 10, 7]

print(notas[0])       # el primero → 8
print(notas[2])       # el tercero → 9
print(notas[-1])      # el último → 7
print(len(notas))     # cuántos elementos hay → 5
````

## 10.2 Modificar una lista

````python
compras = ["leche", "pan"]

compras.append("huevos")     # agregar al final
print(compras)

compras.insert(0, "café")    # agregar en una posición
print(compras)

compras[1] = "leche light"   # cambiar un elemento
print(compras)

compras.remove("pan")        # quitar por valor
print(compras)

ultimo = compras.pop()       # quitar el último (y guardarlo)
print(ultimo, compras)
````

## 10.3 Rebanadas (slicing): sacar un pedazo de la lista
`lista[inicio:fin]` → desde `inicio` hasta `fin`, **sin incluir `fin`** (igual que `range`).

````python
numeros = [10, 20, 30, 40, 50, 60]

print(numeros[1:4])    # [20, 30, 40]
print(numeros[:3])     # los primeros 3 → [10, 20, 30]
print(numeros[3:])     # del índice 3 al final → [40, 50, 60]
print(numeros[::-1])   # al revés → [60, 50, 40, 30, 20, 10]
````

## 10.4 Funciones útiles

````python
numeros = [4, 9, 1, 7, 3]

print("Suma:", sum(numeros))
print("Máximo:", max(numeros))
print("Mínimo:", min(numeros))
print("Ordenada:", sorted(numeros))
print("¿Está el 7?", 7 in numeros)
print("¿Cuántas veces aparece el 9?", numeros.count(9))
````

## 10.5 Listas + `for`: la combinación más usada

````python
notas = [8, 6, 9, 10, 7]

# Recorrer
for nota in notas:
    print(f"Nota: {nota}")

# Promedio con acumulador
suma = 0
for nota in notas:
    suma += nota
print(f"Promedio: {suma / len(notas):.2f}")

# Construir una lista nueva con un for
aprobadas = []
for nota in notas:
    if nota >= 7:
        aprobadas.append(nota)
print("Aprobadas:", aprobadas)
````

````python
# Si necesitas el índice y el valor a la vez: enumerate()
frutas = ["manzana", "plátano", "uva"]
for i, fruta in enumerate(frutas):
    print(f"{i}: {fruta}")
````

# 📘 Clase 11: Funciones

Una **función** es un bloque de código **con nombre** que puedes usar las veces que quieras. Ya usaste muchas: `print()`, `len()`, `input()`, `round()`... Ahora vas a **crear las tuyas**.

**¿Para qué?**
- No repetir el mismo código muchas veces.
- Dividir un problema grande en partes pequeñas.
- Darle un nombre claro a lo que hace un pedazo de código.

```python
def nombre_de_la_funcion(parametros):
    instrucciones
    return resultado
```
- `def` **define** la función (todavía no la ejecuta).
- Los **parámetros** son los datos que recibe.
- `return` **devuelve** el resultado a quien la llamó.

## 11.1 Una función sencilla

````python
def saludar(nombre):
    print(f"¡Hola, {nombre}! Bienvenido.")

saludar("Fabricio")     # así se usa ("se llama") la función
saludar("Ana")
````

## 11.2 `return`: devolver un resultado

````python
def area_rectangulo(base, altura):
    return base * altura

a = area_rectangulo(5, 3)
print(a)
print(area_rectangulo(10, 2.5))
print(f"El área es {area_rectangulo(4, 4)}")
````

### ⚠️ `print` vs. `return`: la confusión más común
- `print` **muestra** algo en pantalla, pero el resultado **no se puede usar después**.
- `return` **devuelve** el resultado para **guardarlo o usarlo** en otros cálculos.

````python
def doble_con_print(n):
    print(n * 2)

def doble_con_return(n):
    return n * 2

x = doble_con_print(5)     # muestra 10...
print("x vale:", x)        # ...pero x vale None (vacío): no devolvió nada

y = doble_con_return(5)    # no muestra nada...
print("y vale:", y)        # ...pero y vale 10 y lo puedo usar
print(y + 1)
````

## 11.3 Funciones con `if` adentro

````python
def es_par(n):
    return n % 2 == 0     # la comparación ya da True o False

print(es_par(10))
print(es_par(7))

def clasificar(numero):
    if numero > 0:
        return "positivo"
    elif numero < 0:
        return "negativo"
    else:
        return "cero"

for n in [5, -3, 0]:
    print(n, "es", clasificar(n))
````

## 11.4 Parámetros con valor por defecto

````python
def precio_con_iva(precio, iva=0.16):
    return precio * (1 + iva)

print(precio_con_iva(100))          # usa el iva por defecto (16%)
print(precio_con_iva(100, 0.08))    # usa el iva que le das (8%)
````

## 11.5 Variables locales
Las variables creadas **dentro** de una función solo existen **dentro** de ella:

````python
def calcular():
    resultado = 42        # variable local
    return resultado

print(calcular())
# print(resultado)        # ← si quitas el #, da NameError: afuera no existe
````

## 12.2 Diccionarios: guardar datos con nombre
Una lista guarda datos por **posición** (0, 1, 2...). Un **diccionario** los guarda por **nombre** (clave). Va entre llaves `{ }` con pares `clave: valor`.

````python
alumno = {
    "nombre": "Fabricio",
    "edad": 20,
    "ciudad": "León",
}

print(alumno["nombre"])          # leer por clave
alumno["edad"] = 21              # cambiar un valor
alumno["carrera"] = "Ingeniería" # agregar una clave nueva
print(alumno)

for clave, valor in alumno.items():   # recorrer
    print(f"{clave}: {valor}")
````

## 12.3 `try` / `except`: que tu programa no truene con errores
Si el usuario escribe letras donde esperabas un número, `int()` da un `ValueError` y el programa se detiene. Con `try` / `except` puedes **atrapar** el error y reaccionar:

````python
texto = "abc"      # imagina que esto lo escribió el usuario

try:
    numero = int(texto)
    print(f"El doble es {numero * 2}")
except ValueError:
    print("Eso no es un número entero")
````

````python
# Combinado con while: pedir hasta que escriba un número válido
while True:
    try:
        edad = int(input("¿Cuántos años tienes? "))
        break                          # si llegó aquí, no hubo error: salimos
    except ValueError:
        print("Escribe solo números, por favor.")

print(f"Tienes {edad} años")
````

## 12.5 Listas por comprensión: crear listas en una línea
Una forma corta de construir una lista con un `for` (y opcionalmente un `if`):

````python
# Forma larga
cuadrados = []
for n in range(1, 11):
    cuadrados.append(n ** 2)
print(cuadrados)

# Forma corta (lista por comprensión): hace lo mismo
cuadrados = [n ** 2 for n in range(1, 11)]
print(cuadrados)

# Con condición: solo los pares
pares = [n for n in range(1, 21) if n % 2 == 0]
print(pares)
````

> 💡 Léela así: *"n al cuadrado **por cada** n en el rango del 1 al 10"*. Si te confunde, usa la forma larga: hacen exactamente lo mismo.
